## 阶段 3 练习：变量生成与数据修改

### 练习 3.1：生成分析变量

把 `annual_ex1` 复制为 `annual_ex3`。在 `annual_ex3` 中生成 `年份` 和 `净利率`。`年份` 来自 `统计截止日期` 的年份部分；`净利率 = 净利润_亿元 / 营业收入_亿元`。显示 `股票简称`、`统计截止日期`、`年份`、`营业收入_亿元`、`净利润_亿元`、`净利率` 六列的前 8 行。

在 `annual_ex3` 中新增 `负债水平`。规则是：`资产负债率 >= 0.7` 记为“较高”，其他记录记为“正常”。先全部赋值为“正常”，再用 `.loc` 把符合条件的行改为“较高”。显示 `股票简称`、`年份`、`资产负债率`、`负债水平` 四列的前 8 行。

### 练习 3.3：生成 2020 年收入排名

从 `annual_ex3` 中筛选 2020 年记录，赋值给变量 `annual_2020_ex3`。按 `营业收入_亿元` 从高到低排序，并生成 `收入排名`。`收入排名` 表示 2020 年公司营业收入从高到低的名次，收入最高为第 1 名。显示 `股票简称`、`营业收入_亿元`、`净利润_亿元`、`收入排名` 四列的前 10 行。

## 阶段 4 练习：常见数据问题处理

### 练习 4.1：清洗证券代码和日期

重新读取 `financial_indicators_dirty.xlsx`，读取时把 `证券代码` 处理成字符串，赋值给变量 `dirty_indicators_ex4`。去掉 `证券代码` 和 `股票简称` 两端空格，把 `证券代码` 补齐为 6 位。把 `统计截止日期` 转成日期类型，并显示转换失败的记录。这里的日期类型表示 pandas 能识别年、月、日，后面可以继续提取年份。

```python
dirty_indicators_ex4 = pd.read_excel(
    "data/financial_indicators_dirty.xlsx",
    converters={"证券代码": str},
)

dirty_indicators_ex4["证券代码"] = dirty_indicators_ex4["证券代码"].str.strip().str.zfill(6)
dirty_indicators_ex4["股票简称"] = dirty_indicators_ex4["股票简称"].str.strip()
dirty_indicators_ex4["统计截止日期"] = pd.to_datetime(dirty_indicators_ex4["统计截止日期"], errors="coerce")

display(dirty_indicators_ex4.loc[dirty_indicators_ex4["统计截止日期"].isna()])
```

### 练习 4.2：清理财务指标数值列

把 `加权平均净资产收益率`、`扣非净资产收益率`、`基本每股收益`、`非经常性损益_亿元`、`扣非净利润_亿元` 转成数值列。处理百分号、逗号、`--`、`缺失` 和空字符串。显示转换后的数据类型，并显示这几列的前 8 行。

```python
number_cols = [
    "加权平均净资产收益率",
    "扣非净资产收益率",
    "基本每股收益",
    "非经常性损益_亿元",
    "扣非净利润_亿元",
]

for col in number_cols:
    text = (
        dirty_indicators_ex4[col]
        .astype("string")
        .str.strip()
        .str.replace(",", "", regex=False)
        .str.replace("%", "", regex=False)
        .replace({"--": pd.NA, "缺失": pd.NA, "": pd.NA})
    )
    dirty_indicators_ex4[col] = pd.to_numeric(text, errors="coerce")

display(dirty_indicators_ex4[number_cols].dtypes)
display(dirty_indicators_ex4[["证券代码", "股票简称"] + number_cols].head(8))
```

### 练习 4.3：删除重复记录和关键缺失

按 `证券代码` 和 `统计截止日期` 删除重复记录，再删除这两个连接键缺失的记录，结果赋值给变量 `indicators_clean_ex4`。用 `扣非净利润_亿元 > 0` 生成 `扣非盈利`，表示扣除非经常性损益后的净利润是否为正。显示清理后的行数、每列缺失值数量，并显示前 5 行。

```python
indicators_clean_ex4 = (
    dirty_indicators_ex4
    .drop_duplicates(subset=["证券代码", "统计截止日期"])
    .dropna(subset=["证券代码", "统计截止日期"])
    .copy()
)
indicators_clean_ex4["扣非盈利"] = indicators_clean_ex4["扣非净利润_亿元"] > 0

print(len(indicators_clean_ex4))
display(indicators_clean_ex4.isna().sum())
display(indicators_clean_ex4.head())
```

## 阶段 5 练习：表的追加与合并

### 练习 5.1：纵向追加年度经营记录

重新读取 `company_annual_operations_clean.xlsx`，读取时把 `证券代码` 处理成 6 位字符串，并把 `股票简称` 重命名为 `证券简称`。把 2018-2019 年记录和 2020 年记录拆成两张表，再用 `concat()` 追加回一张表，赋值给变量 `annual_panel_ex5`。显示 `证券代码`、`证券简称`、`统计截止日期`、`营业收入_亿元` 的前 5 行和后 5 行。

```python
annual_for_merge_ex5 = pd.read_excel(
    "data/company_annual_operations_clean.xlsx",
    converters={"证券代码": lambda x: str(x).strip().zfill(6)},
).rename(columns={"股票简称": "证券简称"})

annual_2018_2019_ex5 = annual_for_merge_ex5.loc[annual_for_merge_ex5["统计截止日期"].dt.year.isin([2018, 2019])]
annual_2020_ex5 = annual_for_merge_ex5.loc[annual_for_merge_ex5["统计截止日期"].dt.year == 2020]
annual_panel_ex5 = pd.concat([annual_2018_2019_ex5, annual_2020_ex5], ignore_index=True)

display(annual_panel_ex5[["证券代码", "证券简称", "统计截止日期", "营业收入_亿元"]].head())
display(annual_panel_ex5[["证券代码", "证券简称", "统计截止日期", "营业收入_亿元"]].tail())
```

### 练习 5.2：合并清洗后的指标表

把 `annual_panel_ex5` 和 `indicators_clean_ex4` 按 `证券代码`、`统计截止日期` 左连接，赋值给变量 `analysis_ex5`。显示 `证券代码`、`证券简称`、`统计截止日期`、`营业收入_亿元`、`基本每股收益` 的前 8 行。

```python
analysis_ex5 = annual_panel_ex5.merge(
    indicators_clean_ex4[["证券代码", "统计截止日期", "基本每股收益", "扣非净利润_亿元", "扣非盈利"]],
    on=["证券代码", "统计截止日期"],
    how="left",
)

display(analysis_ex5[["证券代码", "证券简称", "统计截止日期", "营业收入_亿元", "基本每股收益"]].head(8))
```

### 练习 5.3：比较不同合并方式

构造两张小表 `merge_left_ex5` 和 `merge_right_ex5`。左表包含 3 家公司的 `证券代码`、`证券简称`、`营业收入_亿元`；右表包含其中 2 家公司的 `证券代码` 和 `基本每股收益`，并额外加入 1 家不在左表中的公司。分别用 `how="left"`、`how="inner"`、`how="outer"` 合并这两张小表，展示三个结果。

```python
merge_left_ex5 = pd.DataFrame({
    "证券代码": ["000001", "000002", "000004"],
    "证券简称": ["平安银行", "万科A", "国农科技"],
    "营业收入_亿元": [1535.42, 4191.12, 1.71],
})
merge_right_ex5 = pd.DataFrame({
    "证券代码": ["000001", "000002", "000011"],
    "基本每股收益": [1.40, 3.62, 1.34],
})

display(merge_left_ex5.merge(merge_right_ex5, on="证券代码", how="left"))
display(merge_left_ex5.merge(merge_right_ex5, on="证券代码", how="inner"))
display(merge_left_ex5.merge(merge_right_ex5, on="证券代码", how="outer"))
```

## 阶段 6 练习：分组汇总与组内比较

### 练习 6.1：行业年度摘要表

把 `analysis_ex5` 复制为 `analysis_ex6`，并生成 `年份` 和 `净利率`。`年份` 来自 `统计截止日期` 的年份部分；`净利率 = 净利润_亿元 / 营业收入_亿元`。按 `行业名称` 和 `年份` 分组，计算 `公司数`、`营业收入合计_亿元`、`平均净利率`、`平均资产负债率`，赋值给变量 `industry_year_summary`。其中 `公司数` 是每组不重复公司数量，`营业收入合计_亿元` 是每组营业收入求和，两个平均指标是每组对应列的均值。把分组结果整理成普通表，并显示前 12 行。

```python
analysis_ex6 = analysis_ex5.copy()
analysis_ex6["年份"] = analysis_ex6["统计截止日期"].dt.year
analysis_ex6["净利率"] = analysis_ex6["净利润_亿元"] / analysis_ex6["营业收入_亿元"]

industry_year_summary = (
    analysis_ex6
    .groupby(["行业名称", "年份"])
    .agg(
        公司数=("证券简称", "nunique"),
        营业收入合计_亿元=("营业收入_亿元", "sum"),
        平均净利率=("净利率", "mean"),
        平均资产负债率=("资产负债率", "mean"),
    )
    .reset_index()
)

display(industry_year_summary.head(12))
```

### 练习 6.2：判断公司是否高于行业平均

在 2020 年数据中，用 `groupby()` 和 `transform()` 计算每家公司所在行业的平均营业收入，生成 `行业平均收入_亿元`。再生成 `高于行业平均`。算法是：把公司自己的 `营业收入_亿元` 与 `行业平均收入_亿元` 比较，前者更大则为 `True`。显示 `行业名称`、`证券简称`、`营业收入_亿元`、`行业平均收入_亿元`、`高于行业平均`，并查看前 10 行。

```python
analysis_2020_ex6 = analysis_ex6.loc[analysis_ex6["年份"] == 2020].copy()
analysis_2020_ex6["行业平均收入_亿元"] = analysis_2020_ex6.groupby("行业名称")["营业收入_亿元"].transform("mean")
analysis_2020_ex6["高于行业平均"] = analysis_2020_ex6["营业收入_亿元"] > analysis_2020_ex6["行业平均收入_亿元"]

display(analysis_2020_ex6[["行业名称", "证券简称", "营业收入_亿元", "行业平均收入_亿元", "高于行业平均"]].head(10))
```

## 阶段 7 练习：数据重塑与结果表构造

### 练习 7.1：构造公司收入宽表

用 `pivot_table()` 把 `analysis_ex6` 整理为宽表：以 `证券代码`、`证券简称`、`行业名称` 标识每家公司，年份为列，值为 `营业收入_亿元`。结果赋值给变量 `revenue_wide_ex7`。计算 `收入增长率_2018_2020 = 2020 年营业收入 / 2018 年营业收入 - 1`，显示前 10 行。

```python
revenue_wide_ex7 = analysis_ex6.pivot_table(
    index=["证券代码", "证券简称", "行业名称"],
    columns="年份",
    values="营业收入_亿元",
    aggfunc="first",
)
revenue_wide_ex7["收入增长率_2018_2020"] = revenue_wide_ex7[2020] / revenue_wide_ex7[2018] - 1

display(revenue_wide_ex7.head(10))
```

### 练习 7.2：整理公司结果表

把 `revenue_wide_ex7` 的索引还原成普通列，赋值给变量 `company_summary_ex7`。显示 `证券代码`、`证券简称`、`行业名称`、`2018` 年营业收入、`2020` 年营业收入、`收入增长率_2018_2020` 六列的前 10 行。再把 `company_summary_ex7` 的 `证券代码` 设为索引，并显示前 5 行。

```python
company_summary_ex7 = revenue_wide_ex7.reset_index()
display(company_summary_ex7[["证券代码", "证券简称", "行业名称", 2018, 2020, "收入增长率_2018_2020"]].head(10))

display(company_summary_ex7.set_index("证券代码").head())
```

## 阶段 8 练习：公司年度面板中的时间顺序

### 练习 8.1：计算单家公司收入变化

把 `analysis_ex6` 按 `证券代码` 和 `统计截止日期` 排序，赋值给变量 `panel_ex8`。选出任意一家公司，赋值给变量 `one_company_ex8`。计算 `上一年营业收入_亿元`、`收入增加额_亿元`、`收入增长率`。其中 `上一年营业收入_亿元` 是上一行的营业收入，`收入增加额_亿元 = 本年营业收入 - 上一年营业收入`，`收入增长率 = 本年营业收入 / 上一年营业收入 - 1`。显示该公司的全部年度记录。

```python
panel_ex8 = analysis_ex6.sort_values(["证券代码", "统计截止日期"]).copy()
one_code = panel_ex8["证券代码"].iloc[0]
one_company_ex8 = panel_ex8.loc[panel_ex8["证券代码"] == one_code].copy()
one_company_ex8["上一年营业收入_亿元"] = one_company_ex8["营业收入_亿元"].shift(1)
one_company_ex8["收入增加额_亿元"] = one_company_ex8["营业收入_亿元"].diff()
one_company_ex8["收入增长率"] = one_company_ex8["营业收入_亿元"].pct_change()

display(one_company_ex8[["证券代码", "证券简称", "年份", "营业收入_亿元", "上一年营业收入_亿元", "收入增加额_亿元", "收入增长率"]])
```

### 练习 8.2：在公司内部计算收入增长率

在 `panel_ex8` 中按 `证券代码` 分组，计算每家公司内部的 `上一年营业收入_亿元`、`收入增加额_亿元` 和 `收入增长率`。三个变量的定义和练习 8.1 相同，但必须在每家公司内部计算。筛选 2020 年记录，按 `收入增长率` 从高到低排序，赋值给变量 `top_growth_2020_ex8`。显示 `证券代码`、`证券简称`、`行业名称`、`营业收入_亿元`、`上一年营业收入_亿元`、`收入增加额_亿元`、`收入增长率` 的前 10 行。

```python
panel_ex8["上一年营业收入_亿元"] = panel_ex8.groupby("证券代码")["营业收入_亿元"].shift(1)
panel_ex8["收入增加额_亿元"] = panel_ex8.groupby("证券代码")["营业收入_亿元"].diff()
panel_ex8["收入增长率"] = panel_ex8.groupby("证券代码")["营业收入_亿元"].pct_change()

top_growth_2020_ex8 = panel_ex8.loc[panel_ex8["年份"] == 2020].sort_values("收入增长率", ascending=False)
display(top_growth_2020_ex8[["证券代码", "证券简称", "行业名称", "营业收入_亿元", "上一年营业收入_亿元", "收入增加额_亿元", "收入增长率"]].head(10))
```
